```python
# Gerekli kütüphanelerin içe aktarılması

import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
```

```python
# Giriş ve çıkış değişkenlerinin tanımlanması

oda_sicakligi = ctrl.Antecedent(np.arange(16, 36, 1), 'oda_sicakligi')
dis_ortam_sicakligi = ctrl.Antecedent(np.arange(0, 46, 1), 'dis_ortam_sicakligi')
kullanici_aktivitesi = ctrl.Antecedent(np.arange(0, 11, 1), 'kullanici_aktivitesi')

sogutma_gucu = ctrl.Consequent(np.arange(0, 101, 1), 'sogutma_gucu')
```

```python
# Üyelik fonksiyonlarının tanımlanması

# Oda sıcaklığı üyelik fonksiyonları
oda_sicakligi['cok_soguk'] = fuzz.trimf(oda_sicakligi.universe, [16, 16, 20])
oda_sicakligi['soguk'] = fuzz.trimf(oda_sicakligi.universe, [18, 21, 24])
oda_sicakligi['normal'] = fuzz.trimf(oda_sicakligi.universe, [22, 25, 28])
oda_sicakligi['sicak'] = fuzz.trimf(oda_sicakligi.universe, [26, 29, 32])
oda_sicakligi['cok_sicak'] = fuzz.trimf(oda_sicakligi.universe, [30, 35, 35])

# Dış ortam sıcaklığı üyelik fonksiyonları
dis_ortam_sicakligi['cok_dusuk'] = fuzz.trimf(dis_ortam_sicakligi.universe, [0, 0, 10])
dis_ortam_sicakligi['dusuk'] = fuzz.trimf(dis_ortam_sicakligi.universe, [5, 15, 25])
dis_ortam_sicakligi['orta'] = fuzz.trimf(dis_ortam_sicakligi.universe, [20, 27, 34])
dis_ortam_sicakligi['yuksek'] = fuzz.trimf(dis_ortam_sicakligi.universe, [30, 36, 42])
dis_ortam_sicakligi['cok_yuksek'] = fuzz.trimf(dis_ortam_sicakligi.universe, [38, 45, 45])

# Kullanıcı aktivitesi üyelik fonksiyonları
kullanici_aktivitesi['dusuk'] = fuzz.trimf(kullanici_aktivitesi.universe, [0, 0, 4])
kullanici_aktivitesi['orta'] = fuzz.trimf(kullanici_aktivitesi.universe, [3, 5, 7])
kullanici_aktivitesi['yuksek'] = fuzz.trimf(kullanici_aktivitesi.universe, [6, 10, 10])

# Soğutma gücü üyelik fonksiyonları
sogutma_gucu['cok_dusuk'] = fuzz.trimf(sogutma_gucu.universe, [0, 0, 25])
sogutma_gucu['dusuk'] = fuzz.trimf(sogutma_gucu.universe, [15, 30, 45])
sogutma_gucu['orta'] = fuzz.trimf(sogutma_gucu.universe, [35, 50, 65])
sogutma_gucu['yuksek'] = fuzz.trimf(sogutma_gucu.universe, [55, 70, 85])
sogutma_gucu['cok_yuksek'] = fuzz.trimf(sogutma_gucu.universe, [75, 100, 100])
```

```python
# Üyelik fonksiyonlarının görselleştirilmesi

oda_sicakligi.view()
plt.title("Oda Sıcaklığı Üyelik Fonksiyonları")
plt.show()

dis_ortam_sicakligi.view()
plt.title("Dış Ortam Sıcaklığı Üyelik Fonksiyonları")
plt.show()

kullanici_aktivitesi.view()
plt.title("Kullanıcı Aktivitesi Üyelik Fonksiyonları")
plt.show()

sogutma_gucu.view()
plt.title("Soğutma Gücü Üyelik Fonksiyonları")
plt.show()
```

```python
# Bulanık kural tabanının oluşturulması

kurallar = [

    # Oda çok soğuksa soğutma çok düşük olmalı
    ctrl.Rule(oda_sicakligi['cok_soguk'], sogutma_gucu['cok_dusuk']),

    # Oda soğuksa soğutma düşük veya çok düşük olmalı
    ctrl.Rule(oda_sicakligi['soguk'] & kullanici_aktivitesi['dusuk'], sogutma_gucu['cok_dusuk']),
    ctrl.Rule(oda_sicakligi['soguk'] & kullanici_aktivitesi['orta'], sogutma_gucu['dusuk']),
    ctrl.Rule(oda_sicakligi['soguk'] & kullanici_aktivitesi['yuksek'], sogutma_gucu['dusuk']),

    # Oda normal sıcaklıktaysa enerji tasarrufu öncelikli
    ctrl.Rule(oda_sicakligi['normal'] & dis_ortam_sicakligi['cok_dusuk'], sogutma_gucu['cok_dusuk']),
    ctrl.Rule(oda_sicakligi['normal'] & dis_ortam_sicakligi['dusuk'], sogutma_gucu['dusuk']),
    ctrl.Rule(oda_sicakligi['normal'] & dis_ortam_sicakligi['orta'], sogutma_gucu['dusuk']),
    ctrl.Rule(oda_sicakligi['normal'] & dis_ortam_sicakligi['yuksek'], sogutma_gucu['orta']),
    ctrl.Rule(oda_sicakligi['normal'] & dis_ortam_sicakligi['cok_yuksek'], sogutma_gucu['orta']),

    # Oda sıcaksa dış ortam ve aktiviteye göre soğutma artmalı
    ctrl.Rule(oda_sicakligi['sicak'] & dis_ortam_sicakligi['dusuk'], sogutma_gucu['orta']),
    ctrl.Rule(oda_sicakligi['sicak'] & dis_ortam_sicakligi['orta'], sogutma_gucu['orta']),
    ctrl.Rule(oda_sicakligi['sicak'] & dis_ortam_sicakligi['yuksek'], sogutma_gucu['yuksek']),
    ctrl.Rule(oda_sicakligi['sicak'] & dis_ortam_sicakligi['cok_yuksek'], sogutma_gucu['yuksek']),

    ctrl.Rule(oda_sicakligi['sicak'] & kullanici_aktivitesi['dusuk'], sogutma_gucu['orta']),
    ctrl.Rule(oda_sicakligi['sicak'] & kullanici_aktivitesi['orta'], sogutma_gucu['yuksek']),
    ctrl.Rule(oda_sicakligi['sicak'] & kullanici_aktivitesi['yuksek'], sogutma_gucu['yuksek']),

    # Oda çok sıcaksa soğutma yüksek veya çok yüksek olmalı
    ctrl.Rule(oda_sicakligi['cok_sicak'] & dis_ortam_sicakligi['orta'], sogutma_gucu['yuksek']),
    ctrl.Rule(oda_sicakligi['cok_sicak'] & dis_ortam_sicakligi['yuksek'], sogutma_gucu['cok_yuksek']),
    ctrl.Rule(oda_sicakligi['cok_sicak'] & dis_ortam_sicakligi['cok_yuksek'], sogutma_gucu['cok_yuksek']),

    ctrl.Rule(oda_sicakligi['cok_sicak'] & kullanici_aktivitesi['dusuk'], sogutma_gucu['yuksek']),
    ctrl.Rule(oda_sicakligi['cok_sicak'] & kullanici_aktivitesi['orta'], sogutma_gucu['cok_yuksek']),
    ctrl.Rule(oda_sicakligi['cok_sicak'] & kullanici_aktivitesi['yuksek'], sogutma_gucu['cok_yuksek']),
]
```

```python
# Kontrol sistemi ve simülasyon nesnesinin oluşturulması

klima_kontrol_sistemi = ctrl.ControlSystem(kurallar)
klima_simulasyon = ctrl.ControlSystemSimulation(klima_kontrol_sistemi)

print("Bulanık mantık tabanlı klima kontrol sistemi başarıyla oluşturuldu.")
print("Toplam kural sayısı:", len(kurallar))
```

```python
# Tek bir örnek senaryo ile sistemin test edilmesi

klima_simulasyon.input['oda_sicakligi'] = 30
klima_simulasyon.input['dis_ortam_sicakligi'] = 36
klima_simulasyon.input['kullanici_aktivitesi'] = 7

klima_simulasyon.compute()

print("Oda Sıcaklığı:", 30, "°C")
print("Dış Ortam Sıcaklığı:", 36, "°C")
print("Kullanıcı Aktivitesi:", 7)
print("Hesaplanan Soğutma Gücü:", round(klima_simulasyon.output['sogutma_gucu'], 2))
```

```python
# Farklı giriş senaryoları için sistem çıktılarının hesaplanması

senaryolar = [
    [18, 10, 2],
    [22, 20, 3],
    [25, 28, 5],
    [28, 32, 6],
    [30, 36, 7],
    [32, 40, 8],
    [34, 44, 9],
    [26, 38, 4],
    [29, 25, 8],
    [31, 30, 3]
]

sonuclar = []

for i, senaryo in enumerate(senaryolar, start=1):
    sim = ctrl.ControlSystemSimulation(klima_kontrol_sistemi)

    sim.input['oda_sicakligi'] = senaryo[0]
    sim.input['dis_ortam_sicakligi'] = senaryo[1]
    sim.input['kullanici_aktivitesi'] = senaryo[2]

    sim.compute()

    sogutma = round(sim.output['sogutma_gucu'], 2)

    sonuclar.append([
        i,
        senaryo[0],
        senaryo[1],
        senaryo[2],
        sogutma
    ])

sonuclar
```

```python
import pandas as pd

sonuc_tablosu = pd.DataFrame(
    sonuclar,
    columns=[
        "Senaryo No",
        "Oda Sıcaklığı (°C)",
        "Dış Ortam Sıcaklığı (°C)",
        "Kullanıcı Aktivitesi",
        "Soğutma Gücü (%)"
    ]
)

sonuc_tablosu
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.arange(16, 36, 1)
y = np.arange(0, 46, 1)

X, Y = np.meshgrid(x, y)

Z = np.zeros_like(X, dtype=float)

for i in range(len(y)):
    for j in range(len(x)):
        sim = ctrl.ControlSystemSimulation(klima_kontrol_sistemi)

        sim.input['oda_sicakligi'] = X[i, j]
        sim.input['dis_ortam_sicakligi'] = Y[i, j]
        sim.input['kullanici_aktivitesi'] = 5

        sim.compute()

        Z[i, j] = sim.output['sogutma_gucu']

fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_xlabel('Oda Sıcaklığı')
ax.set_ylabel('Dış Ortam Sıcaklığı')
ax.set_zlabel('Soğutma Gücü')

plt.title("Klima Kontrol Sistemi Surface Viewer")
plt.show()
```
